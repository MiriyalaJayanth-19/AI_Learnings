# Tokenization :(

Tokenization is at the heart of much weirdness of LLMs. Do not brush it off.

- Why can't LLM spell words? **Tokenization**.
- Why can't LLM do super simple string processing tasks like reversing a string? **Tokenization**.
- Why is LLM worse at non-English languages (e.g. Japanese)? **Tokenization**.
- Why is LLM bad at simple arithmetic? **Tokenization**.
- Why did GPT-2 have more than necessary trouble coding in Python? **Tokenization**.
- Why did my LLM abruptly halt when it sees the string "<|endoftext|>"? **Tokenization**.
- What is this weird warning I get about a "trailing whitespace"? **Tokenization**.
- Why the LLM break if I ask it about "SolidGoldMagikarp"? **Tokenization**.
- Why should I prefer to use YAML over JSON with LLMs? **Tokenization**.
- Why is LLM not actually end-to-end language modeling? **Tokenization**.
- What is the real root of suffering? **Tokenization**.

### Final recommendations

- Don't brush off tokenization. A lot of footguns and sharp edges here. Security issues. Safety issues.
- Eternal glory to anyone who can delete tokenization as a required step in LLMs.
- In your own application:
  - Maybe you can just re-use the GPT-4 tokens and tiktoken?
  - If you're training a vocab, ok to use BPE with sentencepiece. Careful with the million settings.
  - Switch to minbpe once it is as efficient as sentencepiece :)

````python
tel = """శ్రీశ్రీ అని పిలవబడే శ్రీరంగం శ్రీనివాసరావు (1910 ఏప్రిల్ 30 - 1983 జూన్ 15) ప్రముఖ తెలుగు కవి, హేతువాది, నాస్తికుడు. విప్లవ కవిగా, సాంప్రదాయ, ఛందోబద్ధ కవిత్వాన్ని ధిక్కరించినవాడిగా, అభ్యుదయ రచయితల సంఘం (అరసం) అధ్యక్షుడిగా, విప్లవ రచయితల సంఘం (విరసం) స్థాపక అధ్యక్షుడిగా అతను ప్రసిద్ధి చెందాడు. మహాప్రస్థానం అతని అత్యంత ప్రజాదరణ పొందిన రచనలలో ఒకటి.
జీవిత గమనం
బాల్యం, విద్యాభ్యాసం: 1910 ఏప్రిల్ 30న పూడిపెద్ది వెంకటరమణయ్య, అప్పలకొండ దంపతులకు జన్మించాడు. శ్రీరంగం సూర్యనారాయణకు దత్తుడగుట వలన ఇతని ఇంటిపేరు శ్రీరంగంగా మారింది. ప్రాథమిక విద్యాభ్యాసం విశాఖపట్నంలో చేసాడు. 1925లో ఎస్.ఎస్.ఎల్.సి. పాసయ్యాడు. 1931లో మద్రాసు విశ్వవిద్యాలయంలో బి.ఏ. (జంతుశాస్త్రం) పూర్తి చేసాడు.

ఉద్యోగ పర్వం: 1935లో విశాఖలోని మిసెస్ ఎ.వి.ఎస్ కాలేజీలో డిమాన్స్ట్రేటరుగా చేరాడు. 1938లో మద్రాసు ఆంధ్ర ప్రభలో సబ్ ఎడిటరుగా చేరి, ఆ తరువాత ఆకాశవాణి (ఢిల్లీ), మిలిటరీ, నిజాం నవాబు వద్ద, ఆంధ్రవాణి పత్రికలో వివిధ ఉద్యోగాలు చేసాడు. 1947లో మద్రాసుకు తిరిగి వచ్చి అక్కడే స్థిరపడ్డాడు.

కుటుంబం: 1925లో వెంకట రమణమ్మతో మొదటి వివాహం జరిగింది. పిల్లలు లేని కారణం చేత 1949లో ఒక బాలికను దత్తత తీసుకున్నాడు. 1956లో సరోజను రెండవ వివాహం చేసుకున్నాడు. రెండవ భార్య ద్వారా ముగ్గురు కుమార్తెలు, ఒక కుమారుడు కలిగారు. వీరి చివరి అమ్మాయి నిడుమోలు మాలా మద్రాసు హైకోర్టు శాశ్వత న్యాయమూర్తిగా విధులు నిర్వర్తించింది.

రాజకీయ, సామాజిక జీవితం: 1955 సార్వత్రిక ఎన్నికలలో కమ్యూనిస్టుల తరఫున చురుగ్గా ప్రచారం చేసాడు (ఆ సమయంలో అనారోగ్యానికి గురై ఆసుపత్రి పాలయ్యాడు). 1969 ప్రత్యేక తెలంగాణ ఉద్యమ సమయంలో వావిలాల గోపాలకృష్ణయ్యతో కలిసి ఖమ్మంలో సమైక్య వాదాన్ని వినిపిస్తూ ఉద్యమకారులను లెక్కచేయకుండా ప్రదర్శన జరిపాడు.

పురస్కారాలు, తుదిశ్వాస: కేంద్ర సాహిత్య అకాడమీ అవార్డు, మొదటి రాజా లక్ష్మీ ఫౌండేషను అవార్డు పొందాడు. 1970లో విశాఖపట్నంలో జరిగిన అతని షష్టిపూర్తి ఉత్సవం సందర్భంగానే విరసం ఏర్పడింది. కొంతకాలం క్యాన్సరు వ్యాధితో పోరాడి 1983 జూన్ 15న మరణించాడు.
తొలి అడుగులు: ఏడవ యేటనే రచనా వ్యాసంగాన్ని ప్రారంభించాడు. ఎనిమిదవ యేట మొదటి గేయాల పుస్తకం ప్రచురితమైంది. కాగితం, సిగరెట్ ప్యాకెట్ వెనుక భాగం.. ఇలా ఏది అందుబాటులో ఉంటే దానిపై రాసేవాడు. 1928లో (తన 18వ ఏట) సాంప్రదాయ పద్ధతిలో ప్రభవ అనే కావ్య సంపుటిని ప్రచురించాడు.
గురజాడ అడుగుజాడ: తరువాతి కాలంలో గ్రాంథిక శైలి, ఛందస్సులను పక్కన పెట్టి వాడుక భాషలో, మాత్రా ఛందస్సులో కవిత్వం రాశాడు
మహాప్రస్థానం: 1933 నుండి 1940 వరకు రాసిన జగన్నాథుని రథచక్రాలు, బాటసారి, భిక్షువర్షీయసి, గర్జించు రష్యా వంటి గొప్ప కవితల సంకలనమే మహాప్రస్థానం. 1950లో మొదటిసారి ప్రచురితమైన ఈ పుస్తకం తెలుగు సాహిత్యపు దశనూ, దిశనూ మార్చి శ్రీశ్రీని మహాకవిని చేసింది. ఇవన్నీ మార్క్సిజం దృక్పథంతో రాసినవే అయినా, రాసేనాటికి తనకు మార్క్సిజం గురించి తెలియదని 1981 లండన్ ముద్రణ ముందుమాటలో అతను స్వయంగా పేర్కొన్నాడు.
ఇతర రచనలు: మరోప్రస్థానం, ఖడ్గ సృష్టి (కవితా సంకలనాలు), చరమరాత్రి (కథల సంపుటి), మరియు ఎన్నో రేడియో నాటికలు
చలనచిత్ర రం
ప్రవేశం: ఆధునిక కవి కావడంతో మద్రాసులో సినీ ప్రముఖులతో ఏర్పడిన పరిచయం 1950లో ప్రత్యక్ష సంబంధంగా మారింది. తెలుగులో మొట్టమొదటి డబ్బింగ్ సినిమా ఆహుతి (హిందీ నీరా ఔర్ నందా అనువాదం) కి మాటలు, పాటలు రాశాడు.
కెరీర్: రోహిణి సంస్థ (హెచ్.ఎం.రెడ్డి) నెలకు 300 రూపాయల జీతంతో ఆస్థాన రచయితగా నియమించుకుంది. నిర్దోషి, ప్రపంచం సినిమాలకు పనిచేశాడు. మైసూరులో బి.విఠలాచార్య నిర్మించిన కన్నడ డబ్బింగ్ సినిమా కన్యాదానం కోసం ఒక్కరోజులో 12 పాటలు రాసి ప్రపంచ చలనచిత్ర లోకంలో రికార్డు సృష్టించాడు.
విజయం, ప్రయోగాలు: నేరుగా తీసిన ఎన్నో చిత్రాలకు పాటలు రాశాడు. అల్లూరి సీతారామరాజు సినిమాలోని తెలుగు వీర లేవరా అతని ఆణిముత్యాల్లో ఒకటి. స్వయంగా చెవిలో రహస్యం అనే డబ్బింగ్ సినిమా తీసి నష్టపోయాడు. ఉషశ్రీ పిక్చర్స్ స్థాపించి తీయాలనుకున్న రుక్మిణీ కళ్యాణం సఫలం కాలేదు. రెండవ భార్య సరోజతో కలిసి సినిమాలకు మాటలు రాసాడు.
వ్యక్తిత్వం శ్రీశ్రీ వ్యక్తిత్వంలో ఎన్నో విరుధ్ధమైన భావాలు, విచిత్రమైన సంఘర్షణలు కనిపిస్తాయి. అతను మొత్తంగా బహిర్ముఖుడు. తీవ్రవిమర్శలకు, పసితనపు మాటలకు సమంగా ప్రసిద్ధుడు. సిద్ధాంతాల గురించి, తోటి కవుల గురించి అతను అభిప్రాయాలు అత్యంత వేగంగా, అతితక్కువ హేతుబద్ధంగా మారుతూండేవి. ఉద్యోగాల్లో ఇమడలేకపోవడం, మొదటి వివాహంలో పిల్లలు కలగకపోవడం, చివరి దశలో దాదాపు 50 ఏళ్ళ వయసు దగ్గరపడ్డాకే రెండో భార్యతో పిల్లలు పుట్టడం, సినిమాల్లో సంపాదించి మొత్తం కోల్పోవడం, తన అస్థిరత వల్ల సాహితీసంఘాల్లో వివాదాలు రావడం ఇలా ఎన్నెన్నో ఒడిదుడుకులు అతని జీవితాన్ని తాకాయి. వాస్తవజీవితంలో అతను సమస్త సాంప్రదాయిక పద్ధతులకూ కట్టుబడ్డాడు. కానీ తన విప్లవభావాలతో వాటినెప్పుడూ వ్యతిరేకిస్తుండేవాడు.
బర్మా పోరాటం అనేది ఆగ్నేయాసియా యుద్ధ రంగంలో భాగంగా బ్రిటిష్ కాలనీ అయిన బర్మాలో జరిగిన వరుస యుద్ధాలు. ఇది ప్రాథమికంగా మిత్రరాజ్యాల దళాలు జపాన్ సామ్రాజ్యపు ఆక్రమణ దళాలకు వ్యతిరేకంగా చేసిన పోరాటం. సామ్రాజ్యవాద జపాన్‌కు థాయ్ ఫయాప్ ఆర్మీ, అలాగే బర్మా రాష్ట్రం, ఆజాద్ హింద్ (స్వేచ్ఛా భారతదేశ తాత్కాలిక ప్రభుత్వం) అనే రెండు స్వాతంత్ర్య ఉద్యమాలు మద్దతు ఇచ్చాయి. ఆక్రమిత ప్రాంతాలలో నామమాత్రపు స్వతంత్ర కీలుబొమ్మ రాష్ట్రాలు స్థాపించబడ్డాయి. కొన్ని భూభాగాలు థాయ్‌లాండ్‌చే చేర్చబడ్డాయి. 1942, 1943 లో బ్రిటిష్ ఇండియాలోని అంతర్జాతీయ మిత్రరాజ్యాల దళం కోల్పోయిన భూభాగాలను తిరిగి స్వాధీనం చేసుకోవడానికి అనేక విఫల దాడులను ప్రారంభించింది. 1944లో పోరాటం తీవ్రమైంది, బ్రిటీష్ సామ్రాజ్య దళాలు సుమారు 1 మిలియన్ భూ, వాయు దళాలతో గరిష్ట స్థాయికి చేరుకున్నాయి. ఈ దళాలు ప్రాథమికంగా బ్రిటిష్ ఇండియా నుండి తీసుకోబడ్డాయి, బ్రిటీష్ ఆర్మీ దళాలు 1,00,000 తూర్పు, పశ్చిమ ఆఫ్రికా వలస దళాలు, అనేక ఇతర డొమినియన్లు, కాలనీల నుండి తక్కువ సంఖ్యలో భూ, వాయు దళాలు ఉన్నాయి.[8] ఈ అదనపు దళాలు 1945లో మిత్రరాజ్యాలు బర్మాను తిరిగి స్వాధీనం చేసుకోవడానికి అనుమతించాయి.
ఈ పోరాటం అనేక ముఖ్యమైన లక్షణాలను కలిగి ఉంది. వాతావరణం, వ్యాధి, భూభాగం సైనిక కార్యకలాపాలపై పెద్ద ప్రభావాన్ని చూపాయని ఈ ప్రాంత భౌగోళిక లక్షణాలు అర్థం. రవాణా మౌలిక సదుపాయాల లేమి కారణంగా దళాలను తరలించడానికి, సరఫరా చేయడానికి, గాయపడినవారిని తరలించడానికి సైనిక ఇంజనీరింగ్, వాయు రవాణాపై ప్రాధాన్యత ఇవ్వబడింది. బ్రిటీష్, యునైటెడ్ స్టేట్స్, చైనీస్ వేర్వేరు వ్యూహాత్మక ప్రాధాన్యతలను కలిగి ఉన్నందున, ఈ ప్రచారం రాజకీయంగా సంక్లిష్టంగా ఉంది. ఆసియా-పసిఫిక్ ప్రాంతంలో, పశ్చిమ మిత్రరాజ్యాలు నిర్వహించిన ఏకైక భూగర్భ సైనిక చర్య ఇది. ఇది యుద్ధం ప్రారంభం నుండి ముగింపు వరకు నిరంతరాయంగా కొనసాగింది. దీనికి ప్రధాన కారణం ఈ ప్రాంత భౌగోళిక స్థానమే. ఇది ఆగ్నేయాసియా నుండి భారతదేశం వరకు విస్తరించి ఉంది. ఈ ప్రచార ప్రాంతంలో యుద్ధం ప్రారంభంలో బ్రిటీష్ వారు కోల్పోయిన భూభాగాలు ఉండటమే కాకుండా, జపాన్ పురోగతిని చివరికి అడ్డుకున్న భారతీయ భూభాగాలు కూడా ఉన్నాయి. ఈ ప్రాంత వాతావరణం ప్రధానంగా రుతుపవనాలపై ఆధారపడి ఉంటుంది. దీని వల్ల సంవత్సరంలో కేవలం సగం కాలం మాత్రమే ప్రభావవంతమైన సైనిక చర్యలు సాధ్యమయ్యాయి. రుతుపవనాలతో పాటు, బ్రిటీష్ ఇండియాలో కరువు, అల్లర్లు వంటి ఇతర కారణాలు కూడా ఈ పోరాటాన్ని పొడిగించాయి. మిత్రరాజ్యాలు కూడా నాజీ జర్మనీ ఓటమికే ప్రాధాన్యత ఇచ్చాయి. ఈ కారణాల వల్ల బర్మా పోరాటం నాలుగు దశలుగా విభజించబడింది. మొదటి దశ జపాన్ దండయాత్ర. ఈ దండయాత్ర ఫలితంగా 1942లో బ్రిటీష్, భారతీయ, చైనా దళాలు వెనక్కి తగ్గాల్సి వచ్చింది. రెండవ దశ 1942 చివర నుండి 1944 ప్రారంభం వరకు కొనసాగింది. ఈ కాలంలో మిత్రరాజ్యాలు బర్మాపై దాడులు చేయాలని విఫలయత్నాలు చేశాయి. మూడవ దశ 1944లో జపాన్ భారతదేశంపై చేసిన దండయాత్ర. ఈ దండయాత్ర ఇంఫాల్, కోహిమా యుద్ధాల తర్వాత చివరికి విఫలమైంది. నాలుగవ, ఆఖరి దశ 1944 చివర నుండి 1945 మధ్యకాలం వరకు కొనసాగింది. ఈ దశలో మిత్రరాజ్యాలు విజయవంతంగా దాడి చేసి బర్మాకు విముక్తి కల్పించాయి.
జపాన్ ఆక్రమించిన ఆగ్నేయాసియా ప్రాంతాలలో చెలరేగిన రాజకీయ వాతావరణం నుండి కూడా ఈ పోరాటం బలంగా ప్రభావితమైంది. జపాన్ పాన్-ఆసియనిస్ట్ విధానాన్ని అనుసరించింది. దండయాత్ర సమయంలో జపాన్ ఒక విప్లవాన్ని స్పాన్సర్ చేసింది, దీని ఫలితంగా బర్మా రాష్ట్రం ఏర్పడింది. ఈ రాష్ట్ర సైన్యం, బర్మా ఇండిపెండెన్స్ ఆర్మీ, దేశానికి వ్యతిరేకంగా జరిగిన దాడులకు నాయకత్వం వహించింది. ముఖ్యంగా 1944లో జరిగిన ఆపరేషన్ యు-గో సమయంలో స్వేచ్ఛా భారతదేశ తాత్కాలిక ప్రభుత్వం, దాని ఇండియన్ నేషనల్ ఆర్మీ జపాన్ సామ్రాజ్యానికి మద్దతుగా పోరాడాయి. ఇండియన్ నేషనల్ ఆర్మీ ఇంతకుముందు నాజీ జర్మనీకి సహకరించింది. బర్మాలో జపాన్ మిలిటరిస్ట్ కమాండర్ల ఆధిపత్య వైఖరి "కో-ప్రాస్పెరిటీ స్ఫియర్" కు తీవ్ర నష్టం కలిగించింది. దీని వల్ల బర్మా ప్రజలకు నిజమైన స్వాతంత్ర్యం వస్తుందన్న ఆశలు సన్నగిల్లాయి, 1945లో బర్మా జాతీయ సైన్యం తిరుగుబాటు చేసింది. మిత్రరాజ్యాల వైపు, యుద్ధంలో ఎక్కువ భాగం రాజకీయ సంబంధాలు మిశ్రమంగా ఉన్నాయి. అమెరికన్లచే శిక్షణ పొందిన చైనీస్ X ఫోర్స్ వల్ల చైనా, అమెరికాల మధ్య సహకారం పెరిగింది. కానీ జనరల్ జోసెఫ్ స్టిల్వెల్, చైనీస్ జనరలిస్సిమో చియాంగ్ కై-షేక్ ప్రతిపాదించిన విరుద్ధమైన వ్యూహాల వల్ల స్టిల్వెల్ అమెరికన్ కమాండర్‌గా తన పదవి నుండి తొలగించబడ్డాడు. మరోవైపు చైనా-భారత సంబంధాలు సానుకూలంగా ఉన్నాయి. చైనీస్ Y ఫోర్స్‌కు చేరుకోవడానికి నిర్మించిన బర్మా రోడ్, హిమాలయాల మీదుగా "ది హంప్" అని పిలువబడే అత్యంత ప్రమాదకరమైన వాయు మార్గం మీదుగా చేసిన మిషన్లు రెండు దేశాల మధ్య సంబంధాలను బలోపేతం చేశాయి. ఈ పోరాటం యుద్ధానంతర సంవత్సరాల్లో బర్మా, భారతదేశ స్వాతంత్ర్య పోరాటంపై గొప్ప ప్రభావాన్ని చూపుతుంది.
మార్చి 1942లో రంగూన్ పతనమైన తర్వాత, మిత్రరాజ్యాలు బర్మా ఉత్తర భాగంలో (ఎగువ బర్మా) తమ స్థానాలను నిలబెట్టుకోవడానికి ప్రయత్నించాయి. వీరికి చైనీస్ ఎక్స్‌పెడిషనరీ ఫోర్స్ సహాయం అందించింది. సింగపూర్‌ను ఆక్రమించిన జపాన్ సైన్యానికి కూడా అదనంగా రెండు విభాగాలు వచ్చి చేరాయి. జపాన్ సైన్యం కొత్తగా ఏర్పడిన బర్మా కార్ప్స్‌ను, చైనా దళాలను ఓడించింది. మిత్రరాజ్యాలు పెరుగుతున్న బర్మీస్ తిరుగుబాటుదారులను కూడా ఎదుర్కోవాల్సి వచ్చింది. వారు ఆధీనంలో ఉంచుకున్న ప్రాంతాల్లో పౌర పరిపాలన కుప్పకూలింది. దాదాపు అన్ని సరఫరా మార్గాలను కోల్పోవడంతో, చివరకు మిత్రరాజ్యాల కమాండర్లు బర్మా నుండి దళాలను వెనక్కి తీసుకోవాలని నిర్ణయించుకున్నారు. ఏప్రిల్ 16న, బర్మాలో యెనాంగ్యాంగ్ యుద్ధం సమయంలో జపనీస్ 33వ విభాగం 7,000 మంది బ్రిటిష్ సైనికులను చుట్టుముట్టింది, వారిని చైనీస్ 38వ విభాగం రక్షించింది.[39]

తిరోగమనం చాలా కష్టమైన పరిస్థితుల్లో జరిగింది. ఆకలితో ఉన్న శరణార్థులు, క్రమశిక్షణ లేని సైనికులు, అనారోగ్యంతో, గాయాలతో ఉన్నవారు భారతదేశానికి వెళ్లే ఆదిమ రహదారులు, బాటలను నింపేశారు. మే 1942లో రుతుపవనాలు ప్రారంభమయ్యే ముందు బర్మా కార్ప్స్ ఎలాగోలా భారతదేశంలోని మణిపూర్‌లోని ఇంఫాల్‌ వరకు చేరుకోగలిగింది. కానీ ఈ ప్రయాణంలో వారు తమ చాలా పరికరాలను, రవాణా వాహనాలను కోల్పోయారు. అక్కడ వారు భారీ వర్షాలలో బహిరంగ ప్రదేశంలో నివసించవలసి వచ్చింది. వాతావరణం, పరిస్థితుల కారణంగా అది చాలా అనారోగ్యకరమైన వాతావరణం. సైనికులు, శరణార్థులకు సహాయం చేయడంలో భారతీయ సైన్యం, పౌర అధికారులు చాలా నెమ్మదిగా స్పందించారు.

కమ్యూనికేషన్ లేకపోవడం వల్ల బ్రిటీష్ వారు బర్మా నుండి ఉపసంహరించుకున్నప్పుడు, చైనీస్ సైనికులకు ఆ విషయం దాదాపు ఎవరికీ తెలియదు. బ్రిటీష్ మద్దతు లేకుండా తాము గెలవలేమని గ్రహించిన చియాంగ్ కై-షేక్ నేతృత్వంలోని X ఫోర్స్ సైనికులు కొందరు హడావిడిగా, క్రమశిక్షణ లేకుండా భారతదేశానికి పారిపోయారు. అక్కడ వారిని అమెరికన్ జనరల్ జోసెఫ్ స్టిల్వెల్ కమాండ్‌లోకి తీసుకున్నారు. కోలుకున్న తర్వాత, వారికి అమెరికన్ ఇన్స్ట్రక్టర్లచే కొత్త ఆయుధాలు, శిక్షణ అందించబడ్డాయి. మిగిలిన చైనా సైనికులు మారుమూల పర్వత అడవుల గుండా యున్నాన్‌కు తిరిగి రావడానికి ప్రయత్నించారు, వీరిలో కనీసం సగం మంది మరణించారు.

బర్మాలోకి థాయ్ ఆర్మీ ప్రవేశం
డిసెంబర్ 21, 1941న జపాన్‌తో థాయిలాండ్ సైనిక కూటమి ఒప్పందం చేసుకుంది. ఆ తర్వాత మార్చి 21న కరెన్నీ రాష్ట్రం, షాన్ రాష్ట్రాలు థాయ్ నియంత్రణలో ఉండాలని థాయ్, జపాన్ దేశాలు అంగీకరించాయి. బర్మా మిగిలిన భాగం జపనీయుల ఆధీనంలోకి వెళ్ళింది.

జనరల్ J. R. సెరిరోయెంగ్రిట్ నాయకత్వంలోని థాయ్ ఫాయప్ సైన్యం మే 10, 1942న సరిహద్దు దాటి షాన్ రాష్ట్రాల్లోకి ప్రవేశించింది. సాయుధ నిఘా దళాలు, రాయల్ థాయ్ వైమానిక దళం మద్దతుతో, మూడు థాయ్ పదాతిదళ విభాగాలు, ఒక అశ్వికదళ విభాగం తిరోగమిస్తున్న చైనీస్ 93వ విభాగాన్ని ఎదుర్కొన్నాయి. మే 27న వారు తమ ప్రధాన లక్ష్యమైన కెంగ్టుంగ్‌ను స్వాధీనం చేసుకున్నారు. జూలై 12న, యుద్ధం తర్వాత ఆక్రమిత షాన్ రాష్ట్రానికి సైనిక గవర్నర్ అయిన జనరల్ ఫిన్ చూన్‌హావన్, ఫాయప్ సైన్యానికి చెందిన 3వ విభాగాన్ని దక్షిణ షాన్ రాష్ట్రం నుండి కరెన్నీ రాష్ట్రాన్ని ఆక్రమించి, లోయికావ్ నుండి చైనీస్ 55వ విభాగాన్ని తరిమికొట్టాలని ఆదేశించారు. యున్నాన్‌కు వెళ్లే మార్గాలు యాక్సిస్ దళాల నియంత్రణలో ఉండటంతో చైనా దళాలు వెనక్కి తగ్గలేకపోయాయి. దీంతో చాలా మంది చైనీస్ సైనికులు బందీలుగా పట్టుబడ్డారు. యుద్ధం ముగిసే వరకు షాన్ రాష్ట్రాలు థాయ్ నియంత్రణలోనే ఉన్నాయి. వారి దళాలు సరఫరా కొరత, వ్యాధులతో బాధపడ్డాయి కానీ మిత్రరాజ్యాల దాడులకు గురికాలేదు.

మిత్రరాజ్యాల ఎదురుదెబ్బలు, 1942–1943

1943లో మయన్మార్‌లోని తముకు వెళ్లే మార్గంలో భారతీయ దళాలు తీవ్ర బురద పరిస్థితుల మధ్య మందుగుండు సామగ్రిని తరలిస్తున్న దృశ్యం.
రుతుపవనాలు ముగిసిన తర్వాత జపాన్ సైన్యం దాడిని పునరుద్ధరించలేదు. బదులుగా బర్మాలో బా మావ్ నేతృత్వంలో ఒక నామమాత్రపు స్వతంత్ర ప్రభుత్వాన్ని ఏర్పాటు చేశారు. బర్మా ఇండిపెండెన్స్ ఆర్మీని జనరల్ ఆంగ్ సాన్ నాయకత్వంలో బర్మా నేషనల్ ఆర్మీగా పునర్వ్యవస్థీకరించారు. అయితే వాస్తవానికి ప్రభుత్వంతో పాటు సైన్యం రెండూ జపనీయుల కఠిన నియంత్రణలోనే ఉన్నాయి.

మిత్రరాజ్యాల విషయానికి వస్తే, 1942, 1943 కాలంలో బర్మాలో వారు చేపట్టిన ఆపరేషన్లు చాలావరకు నిరాశనే మిగిల్చాయి. బ్రిటన్ ఒకేసారి కేవలం మూడు పోరాటాలను మాత్రమే నిర్వహించగలిగే స్థితిలో ఉంది. వనరుల కొరత కారణంగా మధ్యప్రాచ్యం, ఫార్ ఈస్ట్ (తూర్పు ఆసియా) లలో ఒకేసారి దాడులు చేపట్టడం అసాధ్యంగా మారింది. భౌగోళికంగా దగ్గరగా ఉండటం వల్ల, అలాగే లండన్, వాషింగ్టన్‌లు "జర్మనీకి మొదటి ప్రాధాన్యత" అనే విధానాన్ని అవలంబించడం వల్ల మధ్యప్రాచ్యానికే ఎక్కువ ప్రాముఖ్యత లభించింది.

తూర్పు భారతదేశంలో అప్పటి అస్తవ్యస్త పరిస్థితి మిత్రరాజ్యాల సైనిక సమీకరణకు ఆటంకంగా మారింది. బెంగాల్, బీహార్ రాష్ట్రాల్లో 'క్విట్ ఇండియా' ఉద్యమ నిరసనలు తీవ్ర రూపం దాల్చాయి. వాటిని అణచివేయడానికి పెద్ద సంఖ్యలో బ్రిటిష్ సైన్యం అవసరమైంది. ఇదే సమయంలో బెంగాల్‌లో భయంకరమైన కరువు సంభవించింది, దీనివల్ల ఆకలి, వ్యాధుల కారణంగా దాదాపు 30 లక్షల మంది మరణించారు. ఇలాంటి గందరగోళ పరిస్థితుల మధ్య అసోంలోని ముందు వరుస దళాలకు సరఫరా మార్గాలను మెరుగుపరచడం, యుద్ధ అవసరాల కోసం స్థానిక పరిశ్రమలను వినియోగించుకోవడం చాలా కష్టంగా మారింది. మిత్రరాజ్యాల సైనికులకు శిక్షణను మెరుగుపరచే ప్రయత్నాలకు కూడా సమయం పట్టింది. ముందు వరుస ప్రాంతాల్లో పేలవమైన మనోస్థైర్యం, స్థానికంగా వ్యాపించిన వ్యాధుల కారణంగా పోరాట విభాగాల బలం, సామర్థ్యం తగ్గిపోయాయి.

ఇన్ని అడ్డంకులు ఉన్నప్పటికీ, 1942-1943 పొడి కాలంలో (వర్షాకాలం కాని సమయం) మిత్రరాజ్యాలు రెండు ఆపరేషన్లు చేపట్టాయి. అందులో మొదటిది బర్మాలోని తీరప్రాంత అరకాన్ ప్రావిన్స్‌పై చేసిన చిన్న దాడి. భారతీయ తూర్పు సైన్యం (బ్రిటిష్ కమాండ్ కింద ఉన్న) మయూ ద్వీపకల్పం, ముఖ్యమైన ఎయిర్‌ఫీల్డ్ ఉన్న అక్యబ్ ద్వీపాన్ని తిరిగి స్వాధీనం చేసుకోవాలని లక్ష్యంగా పెట్టుకుంది. ఒక విభాగం ద్వీపకల్పం ముగింపుకి కొద్ది మైళ్ల దూరంలో ఉన్న డాన్‌బైక్ వరకు చేరుకుంది, కానీ అక్కడ బాగా స్థిరపడిన చిన్న జపనీస్ దళం వారిని అడ్డుకుంది. ఈ సమయంలో మిత్రరాజ్యాల వద్ద జపనీయుల బలమైన బంకర్‌లను అధిగమించడానికి అవసరమైన ఆయుధాలు, వ్యూహాత్మక సామర్థ్యం లేవు. బ్రిటిష్, భారతీయ దళాలు పదేపదే చేసిన దాడులు విఫలమయ్యాయి, తీవ్ర ప్రాణనష్టం వాటిల్లింది. ఈలోగా మధ్య బర్మా నుండి జపనీస్ అదనపు దళాలు మిత్రరాజ్యాలు అభేద్యమని భావించిన నదులు, పర్వత శ్రేణులను దాటి వచ్చి, మిత్రరాజ్యాల ఎడమ పార్శ్వంపై దాడి చేసి అనేక విభాగాలను ఓడించాయి. అలసిపోయిన బ్రిటిష్ దళాలు ఎలాంటి రక్షణ శ్రేణులను నిలబెట్టుకోలేకపోయాయి, తమ పరికరాలను చాలా వరకు విడిచిపెట్టి భారత సరిహద్దు వరకు వెనక్కి తగ్గవలసి వచ్చింది.

రెండవ చర్య వివాదాస్పదమైనది. బ్రిగేడియర్ ఓర్డే వింగేట్ కమాండ్‌లో, చిండిట్స్ అని పిలువబడే ఒక లాంగ్ రేంజ్ పెనెట్రేషన్ యూనిట్ జపనీస్ ఫ్రంట్ లైన్ల గుండా చొరబడి బర్మాలోకి లోతుగా కవాతు చేసింది. ఈ ఆపరేషన్‌కు 'ఆపరేషన్ లాంగ్‌క్లాత్' అని పేరు పెట్టారు. బర్మాలోని ప్రధాన ఉత్తర-దక్షిణ రైల్వేను కత్తిరించడం దీని ప్రారంభ లక్ష్యం. దాదాపు 3,000 మంది సైనికులు అనేక బృందాలుగా బర్మాలోకి ప్రవేశించారు. వారు ఉత్తర బర్మాలోని జపనీస్ కమ్యూనికేషన్ వ్యవస్థను దెబ్బతీశారు, సుమారు రెండు వారాల పాటు రైల్వేను నిలిపివేశారు. కానీ ఈ పోరాటంలో వారు భారీ ప్రాణనష్టాన్ని చవిచూశారు. ఆపరేషన్ సాధించిన ఫలితాలపై సందేహాలు ఉన్నప్పటికీ, దీనిని ఒక ప్రచార అస్త్రంగా ఉపయోగించారు. ముఖ్యంగా బ్రిటిష్, భారతీయ సైనికులు అడవిలో జపనీయులతో సమానంగా జీవించగలరని, ప్రయాణించగలరని, పోరాడగలరని చాటిచెప్పడానికి దీనిని వినియోగించారు. ఇది మిత్రరాజ్యాల సైనికుల్లో మనోస్థైర్యాన్ని పెంపొందించడానికి ఎంతగానో దోహదపడింది.
"""
eng="""
VERY MOMENT IN BUSINESS happens only once. The next Bill Gates will not
build an operating system. The next Larry Page or Sergey Brin won’t make a
search engine. And the next Mark Zuckerberg won’t create a social network.
If you are copying these guys, you aren’t learning from them.
Of course, it’s easier to copy a model than to make something new. Doing
what we already know how to do takes the world from 1 to n, adding more of
something familiar. But every time we create something new, we go from 0 to
1. The act of creation is singular, as is the moment of creation, and the result
is something fresh and strange.
Unless they invest in the difficult task of creating new things, American
companies will fail in the future no matter how big their profits remain today.
What happens when we’ve gained everything to be had from fine-tuning the
old lines of business that we’ve inherited? Unlikely as it sounds, the answer
threatens to be far worse than the crisis of 2008. Today’s “best practices” lead
to dead ends; the best paths are new and untried.
In a world of gigantic administrative bureaucracies both public and private,
searching for a new path might seem like hoping for a miracle. Actually, if
American business is going to succeed, we are going to need hundreds, or
even thousands, of miracles. This would be depressing but for one crucial
fact: humans are distinguished from other species by our ability to work
miracles. We call these miracles technology.
Technology is miraculous because it allows us to do more with less,
ratcheting up our fundamental capabilities to a higher level. Other animals are
instinctively driven to build things like dams or honeycombs, but we are the
only ones that can invent new things and better ways of making them.
Humans don’t decide what to build by making choices from some cosmic
catalog of options given in advance; instead, by creating new technologies,
we rewrite the plan of the world. These are the kind of elementary truths we
teach to second graders, but they are easy to forget in a world where so much
of what we do is repeat what has been done before.
Zero to One is about how to build companies that create new things. It
draws on everything I’ve learned directly as a co-founder of PayPal and
Palantir and then an investor in hundreds of startups, including Facebook and
SpaceX. But while I have noticed many patterns, and I relate them here, this
book offers no formula for success. The paradox of teaching entrepreneurship
is that such a formula necessarily cannot exist; because every innovation is
new and unique, no authority can prescribe in concrete terms how to be
innovative. Indeed, the single most powerful pattern I have noticed is that
successful people find value in unexpected places, and they do this by
thinking about business from first principles instead of formulas.
This book stems from a course about startups that I taught at Stanford in
2012. College students can become extremely skilled at a few specialties, but
many never learn what to do with those skills in the wider world. My primary
goal in teaching the class was to help my students see beyond the tracks laid
down by academic specialties to the broader future that is theirs to create. One
of those students, Blake Masters, took detailed class notes, which circulated
far beyond the campus, and in Zero to One I have worked with him to revise
the notes for a wider audience. There’s no reason why the future should
happen only at Stanford, or in college, or in Silicon Valley to 1914 was a period of both rapid technological development and rapid
globalization. Between the First World War and Kissinger’s trip to reopen
relations with China in 1971, there was rapid technological development but
not much globalization. Since 1971, we have seen rapid globalization along
with limited technological development, mostly confined to IT.
This age of globalization has made it easy to imagine that the decades
ahead will bring more convergence and more sameness. Even our everyday
language suggests we believe in a kind of technological end of history: the
division of the world into the so-called developed and developing nations
implies that the “developed” world has already achieved the achievable, and
that poorer nations just need to catch up.
But I don’t think that’s true. My own answer to the contrarian question is
that most people think the future of the world will be defined by
globalization, but the truth is that technology matters more. Without
technological change, if China doubles its energy production over the next
two decades, it will also double its air pollution. If every one of India’s
hundreds of millions of households were to live the way Americans already
do—using only today’s tools—the result would be environmentally
catastrophic. Spreading old ways to create wealth around the world will result
in devastation, not riches. In a world of scarce resources, globalization
without new technology is unsustainable.
New technology has never been an automatic feature of history. Our
ancestors lived in static, zero-sum societies where success meant seizing
things from others. They created new sources of wealth only rarely, and in the
long run they could never create enough to save the average person from an
extremely hard life. Then, after 10,000 years of fitful advance from primitive
agriculture to medieval windmills and 16th-century astrolabes, the modern
world suddenly experienced relentless technological progress from the advent
of the steam engine in the 1760s all the way up to about 1970. As a result, we
have inherited a richer society than any previous generation would have been
able to imagine. Any generation excepting our parents’ and grandparents’, that is: in the late
1960s, they expected this progress to continue. They looked forward to a four-day workweek, energy too cheap to meter, and vacations on the moon.
But it didn’t happen. The smartphones that distract us from our surroundings also distract us from the fact that our surroundings are strangely old: only
computers and communications have improved dramatically since midcentury. That doesn’t mean our parents were wrong to imagine a better
future—they were only wrong to expect it as something automatic. Today our challenge is to both imagine and create the new technologies that can
New technology tends to come from new ventures—startups. From the
Founding Fathers in politics to the Royal Society in science to Fairchild
Semiconductor’s “traitorous eight” in business, small groups of people bound
together by a sense of mission have changed the world for the better. The
easiest explanation for this is negative: it’s hard to develop new things in big
organizations, and it’s even harder to do it by yourself. Bureaucratic
hierarchies move slowly, and entrenched interests shy away from risk. In the
most dysfunctional organizations, signaling that work is being done becomes
a better strategy for career advancement than actually doing work (if this
describes your company, you should quit now). At the other extreme, a lone
genius might create a classic work of art or literature, but he could never
create an entire industry. Startups operate on the principle that you need to
work with other people to get stuff done, but you also need to stay small
enough so that you actually can.
Positively defined, a startup is the largest group of people you can convince
of a plan to build a different future. A new company’s most important strength
is new thinking: even more important than nimbleness, small size affords
space to think. This book is about the questions you must ask and answer to
succeed in the business of doing new things: what follows is not a manual or a
record of knowledge but an exercise in thinking. Because that is what a
startup has to do: question received ideas and rethink business from scratch.
PARTY LIKE IT’S 1999
OUR CONTRARIAN QUESTION—What important truth do very few people agree
with you on?—is difficult to answer directly. It may be easier to start with a
preliminary: what does everybody agree on? “Madness is rare in individuals
—but in groups, parties, nations, and ages it is the rule,” Nietzsche wrote
(before he went mad). If you can identify a delusional popular belief, you can
find what lies hidden behind it: the contrarian truth.
Consider an elementary proposition: companies exist to make money, not
to lose it. This should be obvious to any thinking person. But it wasn’t so
obvious to many in the late 1990s, when no loss was too big to be described
as an investment in an even bigger, brighter future. The conventional wisdom
of the “New Economy” accepted page views as a more authoritative, forwardlooking financial metric than something as pedestrian as profit.
Conventional beliefs only ever come to appear arbitrary and wrong in
retrospect; whenever one collapses, we call the old belief a bubble. But the
distortions caused by bubbles don’t disappear when they pop. The internet
craze of the ’90s was the biggest bubble since the crash of 1929, and the
lessons learned afterward define and distort almost all thinking about
technology today. The first step to thinking clearly is to question what we
think we know about the past.
A QUICK HISTORY OF THE ’90S
The 1990s have a good image. We tend to remember them as a prosperous,
optimistic decade that happened to end with the internet boom and bust. But
many of those years were not as cheerful as our nostalgia holds. We’ve long
since forgotten the global context for the 18 months of dot-com mania at
decade’s end.
The ’90s started with a burst of euphoria when the Berlin Wall came down
in November ’89. It was short-lived. By mid-1990, the United States was in
recession. Technically the downturn ended in March ’91, but recovery was
slow and unemployment continued to rise until July ’92. Manufacturing never
fully rebounded. The shift to a service economy was protracted and painful.
1992 through the end of 1994 was a time of general malaise. Images of
dead American soldiers in Mogadishu looped on cable news. Anxiety about
globalization and U.S. competitiveness intensified as jobs flowed to Mexico.
This pessimistic undercurrent drove then-president Bush 41 out of office and
won Ross Perot nearly 20% of the popular vote in ’92—the best showing for
a third-party candidate since Theodore Roosevelt in 1912. And whatever the
cultural fascination with Nirvana, grunge, and heroin reflected, it wasn’t hope
or confidence.
Silicon Valley felt sluggish, too. Japan seemed to be winning the
semiconductor war. The internet had yet to take off, partly because its
commercial use was restricted until late 1992 and partly due to the lack of
user-friendly web browsers. It’s telling that when I arrived at Stanford in
1985, economics, not computer science, was the most popular major. To most
people on campus, the tech sector seemed idiosyncratic or even provincial.
The internet changed all this. The Mosaic browser was officially released in
November 1993, giving regular people a way to get online. Mosaic became
Netscape, which released its Navigator browser in late 1994. Navigator’s
adoption grew so quickly—from about 20% of the browser market in January
1995 to almost 80% less than 12 months later—that Netscape was able to IPO
in August ’95 even though it wasn’t yet profitable. Within five months,
Netscape stock had shot up from $28 to $174 per share. Other tech companies
were booming, too. Yahoo! went public in April ’96 with an $848 million
valuation. Amazon followed suit in May ’97 at $438 million. By spring of
’98, each company’s stock had more than quadrupled. Skeptics questioned
earnings and revenue multiples higher than those for any non-internet
company. It was easy to conclude that the market had gone crazy.
This conclusion was understandable but misplaced. In December
"""
text=tel+eng.lower()
````

````python
def get_stats(ids):
    """
    Count how often each consecutive pair occurs.
    """

    counts = {}

    for pair in zip(ids, ids[1:]):
        counts[pair] = counts.get(pair, 0) + 1

    return counts
````

````python
def merge(ids, pair, new_id):
    """
    Replace every occurrence of pair with new_id.
    """
    new_ids = []
    i = 0
    while i < len(ids):
        if (
            i < len(ids) - 1
            and ids[i] == pair[0]
            and ids[i + 1] == pair[1]
        ):
            new_ids.append(new_id)
            i += 2
        else:
            new_ids.append(ids[i])
            i += 1
    return new_ids
````

````python
class BasicTokenizer:
    def __init__(self):
        self.merges = {}
        self.vocab = {}
    def train(self, text, vocab_size, verbose=False):
        assert vocab_size >= 256
        num_merges = vocab_size - 256
        # Convert text → UTF-8 bytes → integer IDs
        text_bytes = text.encode("utf-8")
        ids = list(text_bytes)

        # Initial vocabulary:
        # 0–255 correspond to raw bytes
        vocab = {
            idx: bytes([idx])
            for idx in range(256)
        }

        merges = {}

        for i in range(num_merges):

            # Count adjacent pairs
            stats = get_stats(ids)

            if not stats:
                break

            # Find most frequent pair
            pair = max(
                stats,
                key=stats.get
            )

            # New token ID
            new_id = 256 + i

            # Merge it
            ids = merge(
                ids,
                pair,
                new_id
            )

            # Remember the merge
            merges[pair] = new_id

            # Build the byte representation of new token
            vocab[new_id] = (
                vocab[pair[0]]
                + vocab[pair[1]]
            )

            if verbose:
                print(
                    f"merge {i + 1}/{num_merges}: "
                    f"{pair} -> {new_id} "
                    f"count={stats[pair]}"
                )

        self.merges = merges
        self.vocab = vocab

    def encode(self, text):
        text_bytes = text.encode("utf-8")
        ids = list(text_bytes)
        while len(ids) >= 2:
            stats = get_stats(ids)
            pair = min(
                stats,
                key=lambda p: self.merges.get(
                    p,
                    float("inf")
                )
            )
            if pair not in self.merges:
                break
            new_id = self.merges[pair]
            ids = merge(
                ids,
                pair,
                new_id)
        return ids

    def decode(self, ids):
        text_bytes = b"".join(
            self.vocab[idx]
            for idx in ids
        )
        return text_bytes.decode(
            "utf-8",
            errors="replace")
````

````python
input_ids = torch.tensor(token_ids[:-1], dtype=torch.long)
target_ids = torch.tensor(token_ids[1:], dtype=torch.long)

print("Input shape:", input_ids.shape)
print("Target shape:", target_ids.shape)

print("\nInput IDs:", input_ids)
print("Target IDs:", target_ids)
````

````python
#gpt configure
vocab_size = len(vocab)
d_model = 32
context_length = 16
len(vocab)
````

````python
token_embedding = nn.Embedding(
    num_embeddings=vocab_size,
    embedding_dim=d_model
)

position_embedding = nn.Embedding(
    num_embeddings=context_length,
    embedding_dim=d_model
)
print("Token embedding shape:", token_embedding.weight.shape)
print("Position embedding shape:", position_embedding.weight.shape)
````

````python
# Use the first 16 tokens for this example
x = input_ids[:context_length]

# Create positions: 0, 1, 2, ..., 15
positions = torch.arange(len(x))

# Look up both types of embeddings
token_vectors = token_embedding(x)
position_vectors = position_embedding(positions)

# Combine them
embedded_input = token_vectors + position_vectors

print("Input token IDs shape:", x.shape)
print("Token vectors shape:", token_vectors.shape)
print("Position vectors shape:", position_vectors.shape)
print("Combined embedding shape:", embedded_input.shape)
````

````python
# Add a batch dimension: (sequence_length, d_model)
x = embedded_input.unsqueeze(0)

print("Attention input shape:", x.shape)
````

````python
attention_dim = d_model

W_q = nn.Linear(d_model, attention_dim, bias=False)
W_k = nn.Linear(d_model, attention_dim, bias=False)
W_v = nn.Linear(d_model, attention_dim, bias=False)

Q = W_q(x)
K = W_k(x)
V = W_v(x)

print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)
````

````python
import math

scores = Q @ K.transpose(-2, -1)
scores = scores / math.sqrt(attention_dim)

print("Attention scores shape:", scores.shape)
````

````python
# True above the diagonal means those positions are in the future
mask = torch.triu(
    torch.ones(context_length, context_length, dtype=torch.bool),
    diagonal=1
)

masked_scores = scores.masked_fill(mask, float("-inf"))

print("Mask shape:", mask.shape)
print("Masked scores shape:", masked_scores.shape)
print("\nFirst 4 rows and columns:")
print(masked_scores[0, :4, :4])
````

````python
attention_weights = torch.softmax(masked_scores, dim=-1)

print("Attention weights shape:", attention_weights.shape)
print("\nFirst 4 rows and columns:")
print(attention_weights[0, :4, :4])
````

````python
attention_output = attention_weights @ V

print("Attention output shape:", attention_output.shape)
````

````python
class CausalSelfAttention(nn.Module):
    def __init__(self, d_model):
        super().__init__()

        self.query = nn.Linear(d_model, d_model, bias=False)
        self.key = nn.Linear(d_model, d_model, bias=False)
        self.value = nn.Linear(d_model, d_model, bias=False)
        self.proj = nn.Linear(d_model, d_model, bias=False)

    def forward(self, x):
        batch_size, seq_len, d_model = x.shape

        Q = self.query(x)
        K = self.key(x)
        V = self.value(x)

        scores = Q @ K.transpose(-2, -1)
        scores = scores / (d_model ** 0.5)

        mask = torch.triu(
            torch.ones(seq_len, seq_len, device=x.device, dtype=torch.bool),
            diagonal=1
        )

        scores = scores.masked_fill(mask, float("-inf"))
        weights = torch.softmax(scores, dim=-1)

        output = weights @ V
        return self.proj(output)


attention = CausalSelfAttention(d_model)

attention_output = attention(x)

print("Input shape:", x.shape)
print("Output shape:", attention_output.shape)
````

````python
class TransformerBlock(nn.Module):
    def __init__(self, d_model):
        super().__init__()

        self.attention = CausalSelfAttention(d_model)

        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)

        self.feed_forward = nn.Sequential(
            nn.Linear(d_model, 4 * d_model),
            nn.GELU(),
            nn.Linear(4 * d_model, d_model)
        )

    def forward(self, x):
        # Self-attention + residual connection
        x = x + self.attention(self.norm1(x))

        # Feed-forward network + residual connection
        x = x + self.feed_forward(self.norm2(x))

        return x


block = TransformerBlock(d_model)
block_output = block(x)

print("Input shape:", x.shape)
print("Transformer block output shape:", block_output.shape)
````

````python
class TinyGPT(nn.Module):
    def __init__(self, vocab_size, d_model, context_length):
        super().__init__()

        self.token_embedding = nn.Embedding(vocab_size, d_model)
        self.position_embedding = nn.Embedding(context_length, d_model)

        self.block = TransformerBlock(d_model)
        self.final_norm = nn.LayerNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)

    def forward(self, token_ids):
        batch_size, seq_len = token_ids.shape

        positions = torch.arange(seq_len, device=token_ids.device)

        x = self.token_embedding(token_ids)
        x = x + self.position_embedding(positions)

        x = self.block(x)
        x = self.final_norm(x)

        logits = self.lm_head(x)
        return logits


model = TinyGPT(
    vocab_size=vocab_size,
    d_model=d_model,
    context_length=context_length
)

print("TinyGPT created.")
print("Total parameters:", sum(p.numel() for p in model.parameters()))
````

````python
# Add a batch dimension to your input sequence
model_input = input_ids[:context_length].unsqueeze(0)

# Run the model
logits = model(model_input)

print("Input shape:", model_input.shape)
print("Logits shape:", logits.shape)
````

````python
train_x = input_ids[:context_length].unsqueeze(0)
train_y = target_ids[:context_length].unsqueeze(0)

print("Training input shape:", train_x.shape)
print("Training target shape:", train_y.shape)
````

````python
import torch.nn.functional as F

logits = model(train_x)

loss = F.cross_entropy(
    logits.reshape(-1, vocab_size),
    train_y.reshape(-1)
)

print("Initial training loss:", loss.item())
````

````python
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001
)

print("Optimizer")
print("Parameters to train:", sum(
    p.numel() for p in model.parameters() if p.requires_grad
))
````

````python
model.train()

optimizer.zero_grad()

logits = model(train_x)

loss = F.cross_entropy(
    logits.reshape(-1, vocab_size),
    train_y.reshape(-1)
)

loss.backward()
optimizer.step()

print("Training loss:", loss.item())
print("One training step completed.")
````

````python
model.eval()

with torch.no_grad():
    logits = model(train_x)

    new_loss = F.cross_entropy(
        logits.reshape(-1, vocab_size),
        train_y.reshape(-1)
    )

print("Loss before first update: 7.1070156")
print("Loss after first update:", new_loss.item())
````

````python

def generate_text(prompt, max_new_tokens=20):
    model.eval()

    # Convert the prompt into token IDs
    prompt_ids = tokenizer.encode(prompt)

    # Keep only the most recent tokens that fit the context
    generated = torch.tensor([prompt_ids[-context_length:]], dtype=torch.long)

    with torch.no_grad():
        for _ in range(max_new_tokens):
            logits = model(generated)

            # Get the next-token predictions from the final position
            next_token_logits = logits[:, -1, :]

            # Select the token with the highest score
            next_token_id = next_token_logits.argmax(dim=-1).item()

            # Append the predicted token
            next_token = torch.tensor([[next_token_id]], dtype=torch.long)
            generated = torch.cat([generated, next_token], dim=1)

            # Keep the context within the model's maximum length
            generated = generated[:, -context_length:]

    # Display the generated token pieces
    generated_ids = generated[0].tolist()
    print("Input:", prompt)
    print("Generated token pieces:")
    print([vocab[token_id][0] for token_id in generated_ids])
````

````python
generate_text("My heart is", max_new_tokens=10)
````

````python
generate_text("నేను", max_new_tokens=10)
generate_text("Hello world", max_new_tokens=10)
generate_text("Artificial intelligence", max_new_tokens=10)
````
